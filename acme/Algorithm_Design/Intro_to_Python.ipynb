## Scripts, interactive work, and arithmetic

Python scripts use the `.py` extension. A module header can explain its purpose, while `if __name__ == "__main__":` keeps demonstration code from running on import. IPython is useful for quick experiments and inspecting objects with `?` or tab completion. Arithmetic uses `+`, `-`, `*`, `/`, `**` for powers, and `%` for remainders. Comparisons and `and`, `or`, `not` form Boolean expressions.

**Problem 1.** Run `python_intro.py` from the command line and check its greeting.

```python
import subprocess
import sys
from pathlib import Path
import importlib.util

script = Path("python_intro.py")
result = subprocess.run([sys.executable, str(script)], capture_output=True, text=True, check=True)
assert result.stdout.splitlines()[0] == "Hello, world!"
print(result.stdout, end="")
spec = importlib.util.spec_from_file_location("python_intro", script)
solutions = importlib.util.module_from_spec(spec)
spec.loader.exec_module(solutions)
```

## Functions and arguments

`def` defines a function. A docstring explains its input and result. `return` passes a value back to the caller; `print` only displays it. Positional arguments follow their declared order, while keyword arguments name the target parameter. `print(..., sep=..., end=...)` controls spacing and line endings.

**Problem 2.** Compute sphere volume using $V=\frac{4}{3}\pi r^3$ with $\pi=3.14159$.

```python
for radius in (0, 1, 3):
    volume = solutions.sphere_volume(radius)
    assert abs(volume - 4 / 3 * 3.14159 * radius**3) < 1e-12
    print(f"radius={radius}: volume={volume:.5f}")
```

**Problem 3.** Print five inputs with five spaces between the first three, then one space between the remaining values. Use `sep` and `end` to control the output.

```python
from contextlib import redirect_stdout
from io import StringIO
buffer = StringIO()
with redirect_stdout(buffer):
    solutions.isolate(1, 2, 3, 4, 5)
assert buffer.getvalue() == "1     2     3 4 5\n"
print("isolate(1, 2, 3, 4, 5) ->", repr(buffer.getvalue()))
```

## Strings, slices, and collections

Strings are immutable sequences. A slice `text[start:stop:step]` selects characters; an omitted bound means the beginning or end. A negative step reverses direction. Lists are mutable, tuples are immutable, sets hold distinct values, and dictionaries map keys to values.

**Problem 4.** `first_half` drops a middle character when the length is odd; `backward` uses a negative slice step.

```python
for text in ("abcdef", "abcde", "a", ""):
    first = solutions.first_half(text)
    reverse = solutions.backward(text)
    assert first == text[:len(text) // 2]
    assert reverse == text[::-1]
    print(f"{text!r}: first_half={first!r}, backward={reverse!r}")
```

**Problem 5.** Start with `bear`, `ant`, `cat`, `dog`; append `eagle`, replace `cat` by `fox`, remove `ant`, sort in reverse order, rename `eagle` to `hawk`, and append `hunter` to the last entry.

```python
items = solutions.list_ops()
assert items == ["fox", "hawk", "dog", "bearhunter"]
print("list_ops() ->", items)
```

## Control flow and iteration

`if`/`elif`/`else` chooses a branch. `while` repeats while a condition holds; `for` iterates over a sequence. `range`, `zip`, `enumerate`, `reversed`, and `sorted` help express iteration. `break` ends a loop and `continue` skips to the next iteration.

**Problem 6.** Apply the lab’s single initial vowel or consonant rule for Pig Latin.

```python
examples = {"apple": "applehay", "banana": "ananabay", "orange": "orangehay", "cat": "atcay"}
for word, expected in examples.items():
    actual = solutions.pig_latin(word)
    assert actual == expected
    print(f"{word} -> {actual}")
```

**Problem 7.** Search products of three-digit factors for the largest palindrome. A palindrome equals its character reversal. Once a decreasing inner loop reaches a product below the best result, it can stop.

```python
largest = solutions.palindrome()
assert largest == 906609
assert str(largest) == str(largest)[::-1]
print("Largest three-digit product palindrome:", largest)
print("Factorization: 913 × 993 =", 913 * 993)
```

## Comprehensions

A list comprehension builds a list from an expression and a loop, with an optional filter. Similar syntax builds sets and dictionaries.

**Problem 8.** Sum the first $n$ terms of the alternating harmonic series, $\sum_{k=1}^{n}(-1)^{k+1}/k$, using a list comprehension.

```python
from math import log
for count in (1, 2, 10, 500_000):
    value = solutions.alt_harmonic(count)
    print(f"n={count:,}: sum={value:.8f}")
assert solutions.alt_harmonic(1) == 1
assert solutions.alt_harmonic(2) == 0.5
assert abs(solutions.alt_harmonic(500_000) - log(2)) < 1.1e-6
print("ln(2) =", f"{log(2):.8f}")
```
